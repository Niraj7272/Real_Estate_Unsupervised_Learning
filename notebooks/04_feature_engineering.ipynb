```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/raw/Housing.csv")

df.head()
```

```python
binary_columns = [
    "mainroad",
    "guestroom",
    "basement",
    "hotwaterheating",
    "airconditioning",
    "prefarea"
]

df[binary_columns] = df[binary_columns].replace({
    "yes": 1,
    "no": 0
})
```

```python
df[binary_columns].head()
```

```python
df = pd.get_dummies(
    df,
    columns=["furnishingstatus"],
    drop_first=True,
    dtype=int
)
```

```python
df.head()
```

```python
features = [
    "price",
    "area",
    "bedrooms",
    "stories",
    "parking",
    "mainroad",
    "guestroom",
    "basement",
    "hotwaterheating",
    "airconditioning",
    "prefarea",
    "furnishingstatus_semi-furnished",
    "furnishingstatus_unfurnished"
]

X = df[features].copy()
```

```python
X.head()
```

```python
X.shape
```

```python
X.isnull().sum()
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)
```

```python
X_scaled.head()
```

```python
df.to_csv("../data/processed/cleaned_housing.csv", index=False)
```

```python
X_scaled.to_csv("../data/processed/scaled_features.csv",index=False)
```
