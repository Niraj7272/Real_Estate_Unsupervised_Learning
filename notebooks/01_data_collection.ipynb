# Real Estate Market Segmentation & Anomaly Detection

## 1. Project objective

The objective of this project is to apply unsupervised machine learning techniques to a real estate dataset.

The project aims to:

- Discover groups of similar properties.
- Compare different clustering algorithms.
- Reduce dimensionality using PCA.
- Detect unusual properties using Isolation Forest.

## 2. Dataset

Dataset: Housing Price Dataset

Source: Kaggle

Learning Type: Unsupervised Learning

## 3. Algorithms

- K-Means Clustering
- Hierarchical Clustering
- DBSCAN
- PCA
- Isolation Forest
